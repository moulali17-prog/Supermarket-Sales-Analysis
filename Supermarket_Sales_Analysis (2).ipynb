## Project objective
Analyze supermarket transaction data to understand revenue patterns, product-line performance, branch performance, payment preferences, customer ratings, and sales by day and hour.

Use the public **Supermarket Sales** dataset from Kaggle:

Download the CSV file from Kaggle, then upload it when prompted in the next section. The original dataset commonly contains 1,000 transactions and 17 columns.

## Tools
Python, Pandas, NumPy, Matplotlib, and Seaborn.

**How to run:** In Google Colab, run each cell from top to bottom. Upload the dataset CSV when prompted. Charts and summary metrics will be generated from the actual data.


```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
import io

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

## 1. Upload and load the dataset
Upload the CSV downloaded from Kaggle. The code uses the uploaded filename, so you do not need to rename it.


```python
uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("Dataset loaded:", file_name)
print("Rows and columns:", df.shape)
display(df.head())
```

## 2. Understand the data

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nDescriptive statistics:")
display(df.describe(include="all").T)
```

## 3. Data quality checks and preparation

```python
print("Missing values per column:")
display(df.isnull().sum().to_frame("Missing values"))

print("Duplicate rows before cleaning:", df.duplicated().sum())

# Remove duplicate rows, if any
df = df.drop_duplicates().copy()

# Convert date/time fields when present
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
    df["Day"] = df["Date"].dt.day_name()
    df["Month"] = df["Date"].dt.month_name()

if "Time" in df.columns:
    df["Time"] = pd.to_datetime(df["Time"], format="%H:%M", errors="coerce")
    df["Hour"] = df["Time"].dt.hour

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("Prepared dataset shape:", df.shape)
display(df.head())
```

## 4. Key performance indicators

```python
if "Total" in df.columns:
    total_sales = df["Total"].sum()
    average_transaction = df["Total"].mean()
    highest_transaction = df["Total"].max()
    lowest_transaction = df["Total"].min()

    print(f"Total sales: {total_sales:,.2f}")
    print(f"Average transaction value: {average_transaction:,.2f}")
    print(f"Highest transaction value: {highest_transaction:,.2f}")
    print(f"Lowest transaction value: {lowest_transaction:,.2f}")
    print(f"Number of transactions: {len(df):,}")
else:
    print("The expected 'Total' column was not found. Check the dataset.")
```

## 5. Sales by product line

```python
if {"Product line", "Total"}.issubset(df.columns):
    product_sales = df.groupby("Product line")["Total"].sum().sort_values(ascending=False)
    display(product_sales.to_frame("Total sales"))

    plt.figure(figsize=(10, 6))
    sns.barplot(x=product_sales.values, y=product_sales.index)
    plt.title("Total Sales by Product Line")
    plt.xlabel("Total sales")
    plt.ylabel("Product line")
    plt.tight_layout()
    plt.show()
else:
    print("Required columns not found.")
```

## 6. Branch performance

```python
if {"Branch", "Total"}.issubset(df.columns):
    branch_sales = df.groupby("Branch")["Total"].sum().sort_values(ascending=False)
    display(branch_sales.to_frame("Total sales"))

    plt.figure(figsize=(7, 5))
    sns.barplot(x=branch_sales.index, y=branch_sales.values)
    plt.title("Total Sales by Branch")
    plt.xlabel("Branch")
    plt.ylabel("Total sales")
    plt.tight_layout()
    plt.show()
```

## 7. Payment method analysis

```python
if "Payment" in df.columns:
    payment_counts = df["Payment"].value_counts()
    display(payment_counts.to_frame("Transactions"))

    plt.figure(figsize=(7, 7))
    plt.pie(payment_counts.values, labels=payment_counts.index, autopct="%1.1f%%", startangle=90)
    plt.title("Transactions by Payment Method")
    plt.tight_layout()
    plt.show()
```

## 8. Customer ratings by product line

```python
if {"Product line", "Rating"}.issubset(df.columns):
    rating_by_product = df.groupby("Product line")["Rating"].mean().sort_values(ascending=False)
    display(rating_by_product.to_frame("Average rating"))

    plt.figure(figsize=(10, 6))
    sns.barplot(x=rating_by_product.values, y=rating_by_product.index)
    plt.title("Average Customer Rating by Product Line")
    plt.xlabel("Average rating")
    plt.ylabel("Product line")
    plt.xlim(0, 10)
    plt.tight_layout()
    plt.show()
```

## 9. Sales by day of the week

```python
if {"Day", "Total"}.issubset(df.columns):
    day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
    daily_sales = df.groupby("Day")["Total"].sum().reindex(day_order).dropna()
    display(daily_sales.to_frame("Total sales"))

    plt.figure(figsize=(10, 5))
    sns.barplot(x=daily_sales.index, y=daily_sales.values)
    plt.title("Total Sales by Day of the Week")
    plt.xlabel("Day")
    plt.ylabel("Total sales")
    plt.xticks(rotation=35)
    plt.tight_layout()
    plt.show()
```

## 10. Sales by hour

```python
if {"Hour", "Total"}.issubset(df.columns):
    hourly_sales = df.groupby("Hour")["Total"].sum().sort_index()
    display(hourly_sales.to_frame("Total sales"))

    plt.figure(figsize=(11, 5))
    sns.lineplot(x=hourly_sales.index, y=hourly_sales.values, marker="o")
    plt.title("Total Sales by Hour")
    plt.xlabel("Hour of day")
    plt.ylabel("Total sales")
    plt.xticks(range(0, 24))
    plt.tight_layout()
    plt.show()
```

## 11. Sales by customer gender

```python
if {"Gender", "Total"}.issubset(df.columns):
    gender_sales = df.groupby("Gender")["Total"].sum().sort_values(ascending=False)
    display(gender_sales.to_frame("Total sales"))

    plt.figure(figsize=(7, 5))
    sns.barplot(x=gender_sales.index, y=gender_sales.values)
    plt.title("Total Sales by Customer Gender")
    plt.xlabel("Gender")
    plt.ylabel("Total sales")
    plt.tight_layout()
    plt.show()
```

## 12. Automated findings summary

```python
print("=" * 55)
print("SUPERMARKET SALES ANALYSIS — SUMMARY")
print("=" * 55)

if "Total" in df.columns:
    print(f"Total sales: {df['Total'].sum():,.2f}")
    print(f"Average transaction value: {df['Total'].mean():,.2f}")
    print(f"Transactions analyzed: {len(df):,}")

if {"Product line", "Total"}.issubset(df.columns):
    p = df.groupby("Product line")["Total"].sum()
    print("Highest-revenue product line:", p.idxmax(), f"({p.max():,.2f})")

if {"Branch", "Total"}.issubset(df.columns):
    b = df.groupby("Branch")["Total"].sum()
    print("Highest-revenue branch:", b.idxmax(), f"({b.max():,.2f})")

if "Payment" in df.columns:
    print("Most frequently used payment method:", df["Payment"].mode().iloc[0])

if {"Product line", "Rating"}.issubset(df.columns):
    r = df.groupby("Product line")["Rating"].mean()
    print("Highest average-rated product line:", r.idxmax(), f"({r.max():.2f})")

if {"Day", "Total"}.issubset(df.columns):
    d = df.groupby("Day")["Total"].sum()
    print("Highest-sales day in this dataset:", d.idxmax(), f"({d.max():,.2f})")

if {"Hour", "Total"}.issubset(df.columns):
    h = df.groupby("Hour")["Total"].sum()
    print("Highest-sales hour in this dataset:", f"{int(h.idxmax())}:00", f"({h.max():,.2f})")

print("=" * 55)
```
